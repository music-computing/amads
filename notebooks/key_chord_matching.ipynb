## <font color='crimson'> 1. Vector rotation solution

```python
def rotate(vector: List[int], n: int) -> List[int]:
    """Cyclically rotate a 12-element pitch-class vector by n semitones.
    Rotating right by n means: the value that was at index i moves to
    index (i + n) mod 12. This models transposing a chord/key up by n
    semitones (n=0 is the identity / root position)."""
    n = n % len(vector)
    if n == 0:
        return list(vector)
    return vector[-n:] + vector[:-n]
```

## <font color='crimson'> 2. Chord-fits-key

```python
def fits_in_key(chord_profile: List[int], key_profile: List[int]) -> bool:
    """
    True if every active pitch class in chord_profile is also active
    in key_profile (chord is a subset of the key's pitch content).
    """
    return all(c <= k for c, k in zip(chord_profile, key_profile))


def rotations_fitting_key(
        chord_profile: List[int],
        key_profile: List[int]
) -> List[int]:
    """Return the list of semitone rotations (0-11) of chord_profile that
    fit entirely inside key_profile. This finds every root note on
    which the given chord quality (e.g. major triad) can be built
    while staying diatonic to the given key."""
    fitting = []
    for n in range(PITCH_CLASSES):
        rotated_chord = rotate(chord_profile, n)
        if fits_in_key(rotated_chord, key_profile):
            fitting.append(n)
    return fitting
```

## <font color='crimson'> 3. Comparing two profiles solution

```python
def shared_pitch_count(profile_a: List[int], profile_b: List[int]) -> int:
    """Number of pitch classes active (1) in both profiles."""
    return sum(1 for a, b in zip(profile_a, profile_b) if a == 1 and b == 1)
```

## <font color='purple'> 4. Heatmap for all 12 major keys

```python
# Set up
NOTE_NAMES = ["C", "C#", "D", "D#", "E", "F",
              "F#", "G", "G#", "A", "A#", "B"]

def build_all_major_keys(key_profile: List[int]) -> List[Tuple[str, List[int]]]:
    """
    Generate the 12 major-key profiles (one per tonic) by
    rotating a single reference major-key profile (assumed rooted on C).
    """
    return [(NOTE_NAMES[n], rotate(key_profile, n)) for n in range(PITCH_CLASSES)]


all_major_keys = build_all_major_keys(major_key)
for name, profile in all_major_keys:
    print(f"{name:>3}: {profile}")
```

## <font color='crimson'> 4. Bonus: similarity heat-map across all 12 major keys (solution)

```python
def build_similarity_matrix(keys: List[Tuple[str, List[int]]]) -> np.ndarray:
    """
    Build an NxN matrix of shared-pitch-class counts between
    every pair of key profiles.
    """
    n = len(keys)
    matrix = np.zeros((n, n), dtype=int)
    for i in range(n):
        for j in range(n):
            matrix[i, j] = shared_pitch_count(keys[i][1], keys[j][1])
    return matrix


similarity_matrix = build_similarity_matrix(all_major_keys)
key_names = [name for name, _ in all_major_keys]
similarity_matrix
```

```python
def plot_similarity_heatmap(matrix: np.ndarray, labels: List[str]) -> None:
    """Render the key-similarity matrix as an annotated heat-map."""
    fig, ax = plt.subplots(figsize=(8, 7))
    im = ax.imshow(matrix, cmap="viridis")

    ax.set_xticks(range(len(labels)))
    ax.set_yticks(range(len(labels)))
    ax.set_xticklabels(labels)
    ax.set_yticklabels(labels)
    ax.set_xlabel("Key")
    ax.set_ylabel("Key")
    ax.set_title("Shared pitch classes between major keys")

    for i in range(len(labels)):
        for j in range(len(labels)):
            ax.text(j, i, matrix[i, j], ha="center", va="center",
                     color="white" if matrix[i, j] < matrix.max() * 0.6 else "black",
                     fontsize=8)

    fig.colorbar(im, ax=ax, label="# shared pitch classes")
    fig.tight_layout()
    plt.show()


plot_similarity_heatmap(similarity_matrix, key_names)
```

Reading the heat-map

- The top-left bottom-right diagonal is always **7**, every key shares all 7 of its own notes with itself.
- Keys a **perfect 5th/4th apart** (e.g. C and G, C and F) share **6** notes, the closest distinct relatives (adjacent on the circle of fifths).
- Keys a **whole tone apart** (e.g. C and D) share **5** notes.
- The **most distant** key (a tritone away, e.g. C and F#) shares the fewest notes (2).
